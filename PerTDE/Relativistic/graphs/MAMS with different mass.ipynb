```python
import numpy as np
import matplotlib.pyplot as plt
import json
from pathlib import Path

# --------------------------------------------------
# Paths
# --------------------------------------------------

fallback_dir = Path(
    "/Users/mchau/mesa_tutorial/PerTDE/Relativistic/Fallback_Data"
)

# --------------------------------------------------
# Models
# --------------------------------------------------

models = {
    r'$1 \ M_{\odot}$':  "MAMS1Msun_z0p0142-m1_rp20",
    r'$3 \ M_{\odot}$':  "MAMS3Msun_z0p0142-m1_rp20",
    r'$5 \ M_{\odot}$': "MAMS5Msun_z0p0142-m1_rp20",
}

# --------------------------------------------------
# Load data
# --------------------------------------------------

pro_fallback_data = {}
pro_energy_data = {}

retro_fallback_data = {}
retro_energy_data = {}

for label, adden in models.items():

    # ==================================================
    # PROGRADE
    # ==================================================

    # Fallback
    t_filename = fallback_dir / f"rel_p_t_{adden}_a0p2.txt"

    print(f"Loading prograde fallback: {t_filename}")

    with open(t_filename, "r") as f:
        t_data = json.load(f)

    pro_fallback_data[label] = {
        k: np.array(v) for k, v in t_data.items()
    }

    # Energy
    E_filename = fallback_dir / f"rel_p_E_{adden}_a0p2.txt"

    print(f"Loading prograde energy:   {E_filename}")

    with open(E_filename, "r") as f:
        E_data = json.load(f)

    pro_energy_data[label] = {
        k: np.array(v) for k, v in E_data.items()
    }

    # ==================================================
    # RETROGRADE
    # ==================================================

    # Change rel_r -> whatever your retrograde prefix is
    t_filename = fallback_dir / f"rel_r_t_{adden}_a0p2.txt"

    print(f"Loading retrograde fallback: {t_filename}")

    with open(t_filename, "r") as f:
        t_data = json.load(f)

    retro_fallback_data[label] = {
        k: np.array(v) for k, v in t_data.items()
    }

    # Energy
    E_filename = fallback_dir / f"rel_r_E_{adden}_a0p2.txt"

    print(f"Loading retrograde energy:   {E_filename}")

    with open(E_filename, "r") as f:
        E_data = json.load(f)

    retro_energy_data[label] = {
        k: np.array(v) for k, v in E_data.items()
    }


# --------------------------------------------------
# Plot style
# --------------------------------------------------

plt.style.use(
    "/Users/mchau/mesa_tutorial/PerTDE/Relativistic/plot_style.mplstyle"
)

plt.rc("text", usetex=True)


# ==================================================
# FIGURE: PROGRADE + RETROGRADE
# ==================================================

fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 12),
    gridspec_kw={
        "wspace": 0.10,
        "hspace": 0.35
    }
)

ax_pro_fallback = axes[0, 0]
ax_pro_energy   = axes[0, 1]

ax_ret_fallback = axes[1, 0]
ax_ret_energy   = axes[1, 1]


# ==================================================
# 1. PROGRADE FALLBACK
# ==================================================

for label, data in pro_fallback_data.items():

    ax_pro_fallback.plot(
        data["x"],
        data["y"],
        label=label
    )

# T^(-5/3) reference
bins_T = np.logspace(0, 6, 200)

ax_pro_fallback.plot(
    bins_T,
    10 * bins_T**(-5/3),
    "k--",
    label=r"$T^{-5/3}$"
)

ax_pro_fallback.set_xscale("log")
ax_pro_fallback.set_yscale("log")

ax_pro_fallback.set_xlabel(r"$T / \Delta T$")
ax_pro_fallback.set_ylabel(
    r"$(\mathrm{d}M/\mathrm{d}T) / (M_\star / \Delta T)$"
)

ax_pro_fallback.set_xlim(1, 5000)
ax_pro_fallback.set_ylim(2e-5, 0.1)

ax_pro_fallback.set_title(
    r"Prograde Fallback Rate",
    fontsize=20,
    pad=12
)

ax_pro_fallback.legend(
    loc="upper left",
    bbox_to_anchor=(0.57, 0.97),
    frameon=True,
    edgecolor="black",
    facecolor="white",
    fontsize=12
)


# ==================================================
# 2. PROGRADE ENERGY DISTRIBUTION
# ==================================================

for label, data in pro_energy_data.items():

    ax_pro_energy.plot(
        data["x"],
        data["y"],
        label=label
    )

ax_pro_energy.set_yscale("log")

ax_pro_energy.set_xlabel(r"$E / \Delta E$")
ax_pro_energy.set_ylabel(
    r"$(\mathrm{d}M/\mathrm{d}E) / (M_\star / \Delta E)$"
)

ax_pro_energy.set_xlim(-0.5, 0.5)
ax_pro_energy.set_ylim(1e-5, None)

ax_pro_energy.set_title(
    r"Prograde Energy Distribution",
    fontsize=20,
    pad=12
)

ax_pro_energy.legend(
    loc="best",
    frameon=True,
    edgecolor="black",
    facecolor="white",
    fontsize=12
)


# ==================================================
# 3. RETROGRADE FALLBACK
# ==================================================

for label, data in retro_fallback_data.items():

    ax_ret_fallback.plot(
        data["x"],
        data["y"],
        label=label
    )

# T^(-5/3) reference
ax_ret_fallback.plot(
    bins_T,
    10 * bins_T**(-5/3),
    "k--",
    label=r"$T^{-5/3}$"
)

ax_ret_fallback.set_xscale("log")
ax_ret_fallback.set_yscale("log")

ax_ret_fallback.set_xlabel(r"$T / \Delta T$")
ax_ret_fallback.set_ylabel(
    r"$(\mathrm{d}M/\mathrm{d}T) / (M_\star / \Delta T)$"
)

ax_ret_fallback.set_xlim(1, 5000)
ax_ret_fallback.set_ylim(2e-5, 0.1)

ax_ret_fallback.set_title(
    r"Retrograde Fallback Rate",
    fontsize=20,
    pad=12
)

ax_ret_fallback.legend(
    loc="upper left",
    bbox_to_anchor=(0.57, 0.97),
    frameon=True,
    edgecolor="black",
    facecolor="white",
    fontsize=12
)


# ==================================================
# 4. RETROGRADE ENERGY DISTRIBUTION
# ==================================================

for label, data in retro_energy_data.items():

    ax_ret_energy.plot(
        data["x"],
        data["y"],
        label=label
    )

ax_ret_energy.set_yscale("log")

ax_ret_energy.set_xlabel(r"$E / \Delta E$")
ax_ret_energy.set_ylabel(
    r"$(\mathrm{d}M/\mathrm{d}E) / (M_\star / \Delta E)$"
)

ax_ret_energy.set_xlim(-0.5, 0.5)
ax_ret_energy.set_ylim(1e-5, None)

ax_ret_energy.set_title(
    r"Retrograde Energy Distribution",
    fontsize=20,
    pad=12
)

ax_ret_energy.legend(
    loc="best",
    frameon=True,
    edgecolor="black",
    facecolor="white",
    fontsize=12
)


# ==================================================
# Layout
# ==================================================

plt.subplots_adjust(
    left=0.09,
    right=0.97,
    bottom=0.08,
    top=0.96,
    wspace=0.10,
    hspace=0.35
)

plt.show()
```
