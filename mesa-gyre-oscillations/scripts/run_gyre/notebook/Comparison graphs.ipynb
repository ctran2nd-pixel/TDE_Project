```python
import h5py
import matplotlib.pyplot as plt
from pathlib import Path

models = {
    "1 M$_\\odot$ ZAMS": Path(
        "/Users/mchau/mesa_tutorial/mesa-gyre-oscillations/"
        "results/run_gyre/ZAMS1Msun_Z0p0142/summary.h5"
    ),
    "1 M$_\\odot$ MAMS": Path(
        "/Users/mchau/mesa_tutorial/mesa-gyre-oscillations/"
        "results/run_gyre/MAMS1Msun_Z0p0142/summary.h5"
    ),
    "1 M$_\\odot$ TAMS": Path(
        "/Users/mchau/mesa_tutorial/mesa-gyre-oscillations/"
        "results/run_gyre/TAMS1Msun_Z0p0142/summary.h5"
    ),
    "1 M$_\\odot$ RGB": Path(
        "/Users/mchau/mesa_tutorial/mesa-gyre-oscillations/"
        "results/run_gyre/RGB1Msun_Z0p0142_logR1p3/summary.h5"
    ),
}

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

for ax, (name, summary_path) in zip(axes.flat, models.items()):

    with h5py.File(summary_path, "r") as f:
        n_pg = f["n_pg"][:]
        omega = f["omega"]["re"][:]

    ax.scatter(n_pg, omega, s=30)

    ax.set_title(name)
    ax.set_xlabel(r"$n_{\rm pg}$")
    ax.set_ylabel(r"Dimensionless angular frequency, $\omega$")
    ax.grid(True, alpha=0.3)

fig.suptitle("GYRE Eigenfrequency Comparison", fontsize=16)

plt.tight_layout()
plt.show()
```
