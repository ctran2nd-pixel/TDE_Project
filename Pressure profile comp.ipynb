```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

ROOT = Path.home() / "mesa_tutorial" / "results" / "mams_profiles"

models = [
    ("ZAMS", ROOT / "ZAMS1Msun_Z0p0142" / "ZAMS1Msun_Z0p0142.profile"),
    ("MAMS", ROOT / "MAMS1Msun_Z0p0142" / "MAMS1Msun_Z0p0142.profile"),
    ("TAMS", ROOT / "TAMS1Msun_Z0p0142" / "TAMS1Msun_Z0p0142.profile"),
    ("RGB", ROOT / "RGB1Msun_Z0p0142" / "RGB1Msun_Z0p0142.profile"),
]

plt.figure(figsize=(8, 6))

for label, file in models:

    d = np.genfromtxt(file, names=True, skip_header=5)

    # Radius normalized by stellar radius
    r = 10**d["logR"]
    r = r / r.max()

    # Pressure profile
    logP = d["logP"]

    plt.plot(r, logP, linewidth=2, label=label)

plt.xlabel(r"$r/R_\star$",)
plt.ylabel(r"$\log_{10}(P)$")
plt.title(r"Pressure Profiles of $1\,M_\odot$ Stellar Models")

plt.legend()

plt.grid(True)
plt.tight_layout()

plt.show()
```
